```python
Home_team_goals = df['FullTimeHomeGoals']
plt.hist(Home_team_goals, bins=  np.arange(-0.5,10.5,1), color='blue', edgecolor='black')
plt.title('Goal distribution on home teams')
plt.xlabel('Goals')
plt.ylabel('Frequency')
plt.xticks(range(0,10,1))
plt.show()
```

```python
Away_team_goals = df['FullTimeAwayGoals']
plt.hist(Away_team_goals, bins=  np.arange(-0.5,10.5,1), color='blue', edgecolor='black')
plt.title('Goal distribution on Away teams')
plt.xlabel('Goals')
plt.ylabel('Frequency')
plt.xticks(range(0,10,1))
plt.show()
```
